# 1. Persiapan Data dan Pemodelan OLS

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# 1. Membuat DataFrame
data = {
    "Partisipan": np.arange(1, 11),
    "Peng Makanan": [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    "Peng Transportasi": [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    "Berat Badan": [60, 55, 65, 70, 62, 75, 80, 85, 90, 95],
}
df = pd.DataFrame(data)

# 2. Memisahkan variabel X dan y
X = df[["Peng Makanan", "Peng Transportasi"]]
y = df["Berat Badan"]

# 3. Membangun model OLS
X_model = sm.add_constant(X)
model = sm.OLS(y, X_model)
results = model.fit()

print("Model OLS Berhasil Dibangun!")
print(f"Jumlah Data (N): {len(df)} baris")
display(df.head())

```

# 2. Ringkasan Hasil Regresi dan Persamaan


```python
# Menampilkan ringkasan hasil regresi[cite: 11, 13]
print("=" * 70)
print("HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())

# Mengambil koefisien dan nilai R-squared[cite: 13]
intercept = results.params["const"]
coef_makanan = results.params["Peng Makanan"]
coef_transportasi = results.params["Peng Transportasi"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# Membuat Persamaan Regresi[cite: 13]
persamaan = (
    f"Y = {intercept:.3f} + ({coef_makanan:.3f}) X1 + ({coef_transportasi:.3f}) X2"
)
print("\n" + "=" * 70)
print("PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"R²         = {r_squared:.4f}")
print(f"Adjusted R² = {adjusted_r_squared:.4f}")
```

# 3. Analisis Data Aktual dan Nilai Prediksi

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# 1. Pastikan data dan model OLS siap kembali
data = {
    'Partisipan': np.arange(1, 11),
    'Peng Makanan': [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    'Peng Transportasi': [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    'Berat Badan': [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(data)

X = df[['Peng Makanan', 'Peng Transportasi']]
y = df['Berat Badan']

X_model = sm.add_constant(X)
model = sm.OLS(y, X_model)
results = model.fit()

# 2. Menghitung nilai prediksi[cite: 14]
df['Prediksi'] = results.predict(X_model)

# 3. Menampilkan tabel data aktual & prediksi[cite: 14]
print("=" * 70)
print("DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(df[['Partisipan', 'Peng Makanan', 'Peng Transportasi', 'Berat Badan', 'Prediksi']].round(2))
```

# 4. Visualisasi 3D Regresi Linear Berganda

```python
import plotly.graph_objects as go

# Membuat grid untuk bidang regresi[cite: 14]
x1 = np.linspace(df["Peng Makanan"].min(), df["Peng Makanan"].max(), 40)
x2 = np.linspace(df["Peng Transportasi"].min(), df["Peng Transportasi"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)
Y_plane = intercept + (coef_makanan * X1) + (coef_transportasi * X2)

# Membuat figure 3D[cite: 14, 15]
fig = go.Figure()

# Menampilkan data aktual sebagai titik scatter[cite: 14]
fig.add_trace(
    go.Scatter3d(
        x=df["Peng Makanan"],
        y=df["Peng Transportasi"],
        z=df["Berat Badan"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Partisipan {p}" for p in df["Partisipan"]],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Peng. Makanan: %{x}<br>"
            "Peng. Transportasi: %{y}<br>"
            "Berat Badan: %{z}<extra></extra>"
        ),
    )
)

# Menampilkan bidang regresi (surface)[cite: 15]
fig.add_trace(
    go.Surface(
        x=X1,
        y=X2,
        z=Y_plane,
        opacity=0.60,
        colorscale="Viridis",
        name="Regression Plane",
        hovertemplate=(
            "Peng. Makanan: %{x:.2f}<br>"
            "Peng. Transportasi: %{y:.2f}<br>"
            "Prediksi Y: %{z:.2f}<extra></extra>"
        ),
    )
)

# Mengatur tata letak (layout) grafik[cite: 15]
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D<br>"
            f"<sup>{persamaan} | R² = {r_squared:.3f}</sup>"
        ),
        "x": 0.5,
    },
    scene={
        "xaxis": {"title": "Pengeluaran Makanan (X1)"},
        "yaxis": {"title": "Pengeluaran Transportasi (X2)"},
        "zaxis": {"title": "Berat Badan (Y)"},
    },
    width=900,
    height=700,
    margin=dict(l=0, r=0, b=0, t=80),
    legend={"x": 0.02, "y": 0.98},
)

# Menampilkan grafik[cite: 15]
fig.show()
```
